## 1. Tải dữ liệu

`data/app_reviews/` có 22 file review, mỗi file một ngân hàng, do `crawlers/app_reviews.py` tải từ Google Play. Tên và ảnh người dùng đã bỏ ngay lúc tải (Nghị định 13/2023/NĐ-CP). Hàm nhận thẳng thư mục và ghép 22 file lại (`pattern` để bỏ qua hai file tổng hợp nằm cùng thư mục).

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data" / "model").exists())
sys.path.insert(0, str(ROOT / "EDA"))
from eda_utils import read_and_explore_data

pd.set_option("display.max_columns", 30)
pd.set_option("display.max_colwidth", 60)
pd.set_option("display.width", 200)

REVIEWS = ROOT / "data" / "app_reviews"   # 22 file, mỗi file một ngân hàng: ACB.csv, BAB.csv, ...
PATTERN = "[A-Z][A-Z][A-Z].csv"            # bỏ qua app_catalog.csv, app_reviews_monthly.csv
```

## 2. Run

### Lần 1: dữ liệu

Khóa của một review là `(app_id, review_id)`.

```python
rv = read_and_explore_data(REVIEWS, pattern=PATTERN, key_columns=["app_id", "review_id"])
```

### Lần 2: nhãn và bỏ cột không cần

 Ở đây nhãn là `score`. Bỏ 4 cột không mang thông tin cho mô hình:
- `source`, `app_role`: hằng số.
- `source_url`, `retrieved_at_utc`: thông tin lúc tải dữ liệu.

Bật `auto_drop_duplicates`.

```python
rv = read_and_explore_data(
    REVIEWS, pattern=PATTERN,
    drop_columns=["source", "app_role", "source_url", "retrieved_at_utc"],
    target="score",
    auto_drop_duplicates=True,
    show=False,
)
rv.shape
```

## 3. Kiểm tra thêm: vì sao thiếu, vì sao trùng

Tìm hiểu **vì sao dữ liệu thiếu** trước khi xử lý, và **đoạn trùng lặp đang kể chuyện gì** (trùng hợp lý hay do lỗi).

```python
print(f"replied_at_utc thiếu: {rv.replied_at_utc.isna().sum():,} · số review CHƯA được trả lời: {(~rv.replied).sum():,}")
print(f"Tỷ lệ review được ngân hàng trả lời: {rv.replied.mean():.1%}")
rv["year"] = pd.to_datetime(rv.at_utc).dt.year
print("\napp_version thiếu theo năm (%):")
print((rv.groupby("year").app_version.apply(lambda s: s.isna().mean() * 100)).round(1).to_string())
print(f"\ncontent trống: {rv.content.fillna('').str.strip().eq('').sum()} review")
print(f"Review trước 2019: {(rv.year < 2019).mean():.1%}")
```

```python
norm = rv.content.fillna("").str.lower().str.replace(r"[^\w\s]", " ", regex=True).str.split().str.join(" ")
dup = norm.duplicated(keep=False)
print(f"Dòng có nội dung trùng với dòng khác: {dup.sum():,} ({dup.mean():.1%}) — nhưng không dòng nào trùng hẳn (khác review_id, thời gian)")
top = (pd.DataFrame({"nội dung": norm, "sao": rv.score})
       .groupby("nội dung").agg(số_lần=("sao", "size"), sao_TB=("sao", "mean"))
       .sort_values("số_lần", ascending=False).head(12))
top.round(2)
```

```python
by_bank = rv.ticker.value_counts()
print(f"Review nhiều nhất: {by_bank.index[0]} {by_bank.iloc[0]:,} ({by_bank.iloc[0] / len(rv):.0%}) · ít nhất: {by_bank.index[-1]} {by_bank.iloc[-1]:,}")
print(f"3 NH nhiều nhất chiếm {by_bank.head(3).sum() / len(rv):.0%} tổng số review")
print(f"Số app theo NH: {rv.groupby('ticker').app_id.nunique().value_counts().sort_index().to_dict()}  (NH có hơn 1 app: "
      f"{rv.groupby('ticker').app_id.nunique().loc[lambda s: s > 1].index.tolist()})")
```

## 4. Ý nghĩa đặc trưng theo nhóm

Ký hiệu như notebook 01: 🎯 nhãn · ⚠️ vấn đề chất lượng · 🔍 giả thuyết cần kiểm.

## 5. Insight ban đầu

**Tổng quan: sạch về cấu trúc.** 532.007 review, 14 cột, 22 NH, 2012–2026.
- Khóa `(app_id, review_id)` là duy nhất, không có dòng trùng hẳn.
- Chỉ 19 review có nội dung trống.

**Kiểu dữ liệu:** ba cột thời gian (`at_utc`, `replied_at_utc`, `retrieved_at_utc`) đang lưu dạng chữ, cần đổi sang datetime trước khi gộp theo tháng hay quý.

**Thiếu có cấu trúc, không cần điền:**
- `replied_at_utc` thiếu 257.164 ô, đúng bằng số review chưa được trả lời.
- `app_version` thiếu khoảng 19%, đều qua các năm: nguồn không ghi, không phải thiếu theo thời kỳ.

**Cột hằng số:** `source` và `app_role` chỉ có một giá trị, không mang thông tin nên đã bỏ.

**Nhãn `score` mất cân bằng nặng.**

| Sao | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|
| Tỷ lệ | 37,0% | 3,8% | 4,0% | 4,6% | 50,6% |

- Lớp nhiều nhất gấp **13 lần** lớp ít nhất.
- Nhóm 2–4 sao chỉ có 12,4%.
- ⇒ Không dùng accuracy mà báo macro-F1, và dùng trọng số lớp.
- ⇒ Cân nhắc gộp thành 3 lớp (tiêu cực, trung tính, tích cực).

**Trùng nội dung là trùng hợp lý, nhưng vẫn gây rò rỉ.**
- 46% dòng có nội dung trùng với dòng khác, chủ yếu là câu một, hai chữ: “ok” xuất hiện 52.640 lần, “tốt” 32.906 lần, gần như toàn 5 sao. Đó là nhiều người cùng viết một câu, không phải lỗi hệ thống.
- nếu cùng một câu vừa nằm ở tập train vừa nằm ở tập test thì mô hình “biết trước đáp án”.
- ⇒ Bỏ trùng trước khi chia train/test.

**Lệch giữa các ngân hàng:** riêng MBB có 172.620 review (32%), ba NH nhiều nhất chiếm 56%, còn PGB chỉ có 353 review. Gộp điểm theo NH phải co rút về trung bình (Bayes), và lấy mẫu gán nhãn phải phân tầng theo NH.

## 6. Tổng kết và bước tiếp theo

**Kết luận EDA sơ bộ: dữ liệu đủ sạch để dựng baseline phân loại cảm xúc.**
- Rủi ro chính là **nhãn lệch hai cực**, **nội dung trùng** và **lệch giữa các ngân hàng**, chứ không phải ô thiếu.

Bước tiếp theo:
1. **EDA chi tiết** (`03_EDA_chi_tiet.ipynb`, phần review): độ dài review so với số sao, cách viết (có dấu, không dấu, tiếng Anh), độ phủ khía cạnh.
2. **Tiền xử lý**: đổi 3 cột thời gian sang datetime, bỏ 19 review trống, bỏ trùng nội dung trước khi chia train/test, chuẩn hóa Unicode.
3. **Baseline**: TF-IDF + hồi quy logistic, báo macro-F1; sau đó mới so với PhoBERT hoặc ViSoBERT.
